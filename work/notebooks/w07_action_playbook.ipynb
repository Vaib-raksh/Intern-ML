## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

##Ranked actions and reason codes

The playbook prioritizes pages for human review using the baseline action score developed earlier. Pages with a larger action score are ranked higher because they represent a larger measured gap between impressions and clicks.

**Action:**
- Optimize page — prioritize the page for human review and possible content improvement.

**Reason code:**
- RC1 — High opportunity based on the observed impressions-to-click gap.

The queue is a decision-support tool, not an automatic recommendation to change a page. The ranking is directional and should be reviewed alongside the page context before any action is taken.

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

### Intended use and limits

This playbook is intended for SEO/content teams as a decision-support tool for prioritizing pages that may deserve human review. The ranked actions use observed search-performance data and the baseline action score to identify pages with larger opportunities for improvement.

The recommendations are directional, not guaranteed outcomes. The model performed only slightly differently from the baseline and the time-aware Random Forest did not show a meaningful improvement, so the output should not be treated as a reliable production prediction system.

The queue should be used to decide **which pages to review first**, not to automatically rewrite, remove, publish, or change content. Human review is required before taking any action.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

# Human review + no-go list

Before acting on a ranked page, a content team member should review the page context, search intent, current content quality, and whether the observed search-performance pattern is stable enough to justify an action. The ranking should be treated as a prioritization signal rather than proof that the page needs a specific change.

Human review is required before changing titles, content, structure, internal links, or other page elements. Reviewers should also consider whether the page has a valid reason for its current performance that is not captured by the available features.

**No-go cases — do not automate:**
- Automatically rewriting or publishing page content.
- Automatically changing titles or metadata.
- Automatically deleting, redirecting, or deindexing pages.
- Automatically making changes based only on a low CTR or high action score.
- Treating the model output as proof that a content change will increase CTR.
- Making client-facing recommendations without human validation.

The playbook is intended to support human decisions, not replace them.

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

## Monitoring / retrain triggers

The recommendations should be reviewed periodically because search performance can change over time.

A refresh or re-evaluation should be triggered when:

- The distribution of impressions, clicks, CTR, or position changes substantially from the data used to build the queue.
- The relationship between position and CTR appears to change from the previously observed pattern.
- The baseline or Random Forest metrics change materially when the analysis is rerun on newer data.
- New features or data sources become available that may provide additional predictive signal.
- The ranked queue begins producing many pages that human reviewers consider irrelevant or low-value.

The model should be retrained or the playbook rebuilt when the underlying search data has materially changed or when monitoring shows that the existing recommendations are no longer useful.

These are review and refresh triggers, not automatic production actions.
